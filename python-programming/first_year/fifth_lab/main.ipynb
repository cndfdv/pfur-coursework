```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from IPython.display import HTML
from matplotlib.animation import FuncAnimation

sns.set(style="whitegrid")
```

### Task 1

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)
z = np.cos(x)

plt.figure(figsize=(16, 9))

plt.plot(x, y, label="sin(x)", color="blue")
plt.plot(x, z, label="cos(x)", color="red")
plt.title("Sine and Cosine Functions")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()
plt.show()
```

### Task 2

```python
x = np.random.normal(loc=0, scale=1, size=1000)

plt.figure(figsize=(16, 9))

plt.hist(x, bins=30, color="purple")
plt.title("Histogram of Random Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.grid()
```

### Task 5

```python
fig, axs = plt.subplots(1, 3, figsize=(16, 4))

# Линейный график
x = np.linspace(0, 10, 100)
axs[0].plot(x, x**2)
axs[0].set_title("y = x^2")

# Точечный график
x_rand = np.random.rand(50)
y_rand = np.random.rand(50)
axs[1].scatter(x_rand, y_rand, color="green")
axs[1].set_title("Случайные точки")

# Столбчатая диаграмма
categories = ["A", "B", "C"]
values = [3, 7, 2]
axs[2].bar(categories, values, color="orange")
axs[2].set_title("Категориальные данные")

plt.tight_layout()
plt.show()
```

### Task 6

```python
x_data = np.linspace(0, 5 * np.pi, 100)
y_data = np.sin(x_data)

fig = plt.figure()
plt.xlim(0, 2 * np.pi)
plt.ylim(-1.1, 1.1)
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.title("sin liveplot")
(line,) = plt.plot([], [], lw=2)


def init():
    line.set_data([], [])
    return (line,)


def update(frame):
    x = x_data[:frame]
    y = y_data[:frame]
    line.set_data(x, y)
    return (line,)


ani = FuncAnimation(
    fig, update, frames=len(x_data) + 1, init_func=init, blit=True, interval=50
)

HTML(ani.to_jshtml())
```
